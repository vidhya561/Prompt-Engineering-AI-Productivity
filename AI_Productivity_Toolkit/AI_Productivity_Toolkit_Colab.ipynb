## 2. Build a structured prompt

```python
def build_prompt(role, context, task, constraints, output_format):
    constraint_text = "\\n".join(f"- {item}" for item in constraints)
    return (
        f"Role: {role}\\n"
        f"Context: {context}\\n"
        f"Task: {task}\\n"
        f"Constraints:\\n{constraint_text}\\n"
        f"Output format: {output_format}"
    )

print(build_prompt(
    "Beginner-friendly Python tutor",
    "The learner is new to programming",
    "Explain Python functions",
    ["Use simple English", "Include one code example", "Give three practice questions"],
    "Definition, code, explanation, questions"
))
```

## 5. Structured JSON output

```python
import json

json_prompt = """
Extract details from this task description.
Return ONLY valid JSON with exactly these keys:
{
  "task_name": "string",
  "skills": ["string"],
  "deliverables": ["string"]
}
Do not invent details.
Text: Module 2 asks for prompt engineering practice, a GitHub repository,
a Google Colab link, and a LinkedIn post link.
"""
print(json_prompt)
```

After asking the model, replace the sample JSON below with its actual response. This example is provided to demonstrate validation.

```python
model_json_response = """
{
  "task_name": "Prompt engineering practice",
  "skills": ["zero-shot prompting", "few-shot prompting", "JSON output"],
  "deliverables": ["GitHub repository", "Google Colab link", "LinkedIn post"]
}
"""

try:
    parsed = json.loads(model_json_response)
    required_keys = {"task_name", "skills", "deliverables"}
    if not isinstance(parsed, dict):
        raise ValueError("Expected a JSON object.")
    missing = required_keys - set(parsed)
    if missing:
        raise ValueError(f"Missing keys: {sorted(missing)}")
    print("Valid JSON; required keys are present.")
    print(json.dumps(parsed, indent=2))
except (json.JSONDecodeError, ValueError) as error:
    print("JSON validation failed:", error)
```
